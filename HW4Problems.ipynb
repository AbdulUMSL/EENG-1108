```python
# P1
import math

points = [(2,3), (5,7), (1,8), (2,3), (9,1), (5,7)]

# 1. Remove duplicates
unique_points = list(set(points))

# 2 & 3. Compute distances and find the farthest pair
max_dist = -1
farthest_pair = ()

for i in range(len(unique_points)):
    for j in range(i + 1, len(unique_points)):
        p1 = unique_points[i]
        p2 = unique_points[j]

        # Euclidean Distance: sqrt((x2-x1)^2 + (y2-y1)^2)
        dist = math.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)

        if dist > max_dist:
            max_dist = dist
            farthest_pair = (p1, p2, dist)

# 4. Return result
print(f"Farthest pair: {farthest_pair}")
```

```python
#P2
import re

text = "Python is great. Python is fast, and learning Python is fun!"

# 1. Clean text: remove punctuation and lowercase
clean_text = re.sub(r'[^\w\s]', '', text).lower()
words = clean_text.split()

# 2. Build frequency dictionary
freq_dict = {}
for word in words:
    freq_dict[word] = freq_dict.get(word, 0) + 1

# 3. Sort and get top 5
sorted_freq = sorted(freq_dict.items(), key=lambda x: x[1], reverse=True)
top_5 = sorted_freq[:5]

print(f"Dictionary: {freq_dict}")
print(f"Top 5: {top_5}")
```

```python
#P3
students = {
    "Alice": [88, 92, 79],
    "Bob": [90, 85, 87],
    "Charlie": [70, 80, 75]
}

def get_letter_grade(score):
    if score >= 90: return 'A'
    elif score >= 80: return 'B'
    elif score >= 70: return 'C'
    elif score >= 60: return 'D'
    return 'F'

# 1 & 2. Compute averages and create new dictionary
gradebook = {}
for name, scores in students.items():
    avg = sum(scores) / len(scores)
    gradebook[name] = (avg, get_letter_grade(avg))

# 3. Sort and find top student
sorted_students = sorted(gradebook.items(), key=lambda x: x[1][0], reverse=True)
top_student = sorted_students[0]

print(f"Gradebook: {gradebook}")
print(f"Top Student: {top_student[0]} with average {top_student[1][0]:.2f}")
```

```python
#P4
inventory = {
    "laptop": {"price": 1200, "quantity": 4},
    "mouse": {"price": 25, "quantity": 20},
    "keyboard": {"price": 75, "quantity": 10}
}

# 1. Total inventory value
def get_total_value(inv):
    return sum(item['price'] * item['quantity'] for item in inv.values())

# 2 & 4. Sorted list by total value (also reveals most valuable)
def get_sorted_inventory(inv):
    val_list = [(prod, info['price'] * info['quantity']) for prod, info in inv.items()]
    return sorted(val_list, key=lambda x: x[1], reverse=True)

# 3. Update quantities
def sell_item(inv, product, amount):
    if product in inv and inv[product]['quantity'] >= amount:
        inv[product]['quantity'] -= amount
    else:
        print(f"Error: Not enough {product} in stock.")

# Execution
sorted_vals = get_sorted_inventory(inventory)
print(f"Total Inventory Value: ${get_total_value(inventory)}")
print(f"Most Valuable Product: {sorted_vals[0][0]}")
print(f"Sorted Values: {sorted_vals}")
```
