```python
def is_leap_year(year):
    # If a year is a multiple of 400, it is a leap year
    if year % 400 == 0:
        return True
    # If a year is a multiple of 100, it is not a leap year
    # (We already checked 400, so any remaining multiples of 100 are not leap years)
    elif year % 100 == 0:
        return False
    # If a year is a multiple of 4, it is a leap year
    elif year % 4 == 0:
        return True
    # Otherwise, it is not a leap year
    else:
        return False

try:
    user_input = int(input("Enter a year to check: "))
    if is_leap_year(user_input):
        print(f"{user_input} is a leap year.")
    else:
        print(f"{user_input} is not a leap year.")
except ValueError:
    print("Please enter a valid numerical year.")
```
